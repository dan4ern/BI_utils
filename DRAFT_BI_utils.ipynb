```python
def filter_fastq(input_fastq,
                 output_fastq,
                 gc_bounds = (0, 100),
                 length_bounds = (0, 2**32),
                 quality_threshold = 0):
    reads = read_fastq(input_fastq = input_fastq)
    filtered_reads = []
    for read in reads:
        if check_gc_content(gc_bounds = gc_bounds, read_from_main = read) and check_length(length_bounds = length_bounds, read_from_main = read) and check_quality_threshold(quality_threshold = quality_threshold, read_from_main = read):
            filtered_reads.append(read)
    write_fastq(filtered_reads = filtered_reads, output_fastq = output_fastq)
    return 'Filtration finished!'
```

```python
def read_fastq(input_fastq):
    with open(input_fastq) as inp_fastq:
        reads = []
        for read_data in inp_fastq:
            read_seq = inp_fastq.readline()
            read_comment = inp_fastq.readline()
            read_quality = inp_fastq.readline()
            read = [read_data, read_seq, read_comment, read_quality]
            reads.append(read)
        return reads
```

```python
def check_gc_content(gc_bound, read_from_main):
    read_gc = (read_from_main[1].count("G") + read_from_main[1].count("C")) / len(read_from_main[1].replace('\n', '')) * 100
    if not isinstance(gc_bound, int) and not isinstance(gc_bound, float):
        return read_gc >= gc_bound[0] and read_gc <= gc_bound[1]
    else:
        return read_gc <= gc_bound
```

```python
def check_quality_threshold(read_from_main, quality_threshold):
    clean_quality_string = read_from_main[3].replace('\n', '')
    length = len(clean_quality_string)
    summ_quality = 0
    for n in clean_quality_string:
        summ_quality += ord(n) - 33
    return summ_quality / length >= quality_threshold
```

```python
def read_fasta(input_fasta):
    with open('example_multiline_fasta.fasta') as inp_fasta:
        seqs = []
        seq = []
        for line in inp_fasta:
            if not line.startswith('>'):
                seq.append(line.replace('\n', ''))
            else:
                if seq:
                    seqs.append(seq)
                seq = []
                seq.append(line.replace('\n', ''))
        if seq:
            seqs.append(seq)
    return seqs
```

```python
def convert_multiline_fasta_to_oneline(input_fasta, output_fasta = None):
    seqs = read_fasta(input_fasta = input_fasta)
    oneline_seqs = []
    for seq_list in seqs:
        header = seq_list[0]
        sequence = ''.join(seq_list[1:])
        oneline_seqs.append([header, sequence])
    write_fasta(fasta_one_line = oneline_seqs, output_fasta = output_fasta)
    return 'Conversion finished!'
```

```python
def parse_blast_output(input_file, output_file):
    blasts = read_blast(input_file = input_file)
    hits = []
    for blast in blasts:
        description_end = blast[1].index('Name')
        first_hit = blast[2][:description_end].strip()
        hits.append(first_hit)
    hits.sort()
    write_blast(output_file = output_file, hits = hits)
```

```python
def read_blast(input_file):
    blasts = []
    with open('example_blast_results.txt') as inp_blast:
        for line in inp_blast:
            if line.startswith('Query #'):
                for _ in range(3):
                    inp_blast.readline()
                columns = inp_blast.readline()
                first_blast = inp_blast.readline()
                blasts.append([line, columns, first_blast])
    return blasts
```

```python
def write_blast(output_file, hits):
    if os.path.isfile(output_file):
        raise Exception('Output file with this name already exists. Choose another name.')
    with open(output_file, mode = 'w') as out_blast:
        for hit in hits:
            out_blast.write(hit + '\n')
    return 'BLAST parsing finished!'
```
